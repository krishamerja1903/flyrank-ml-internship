## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/krishamerja1903/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

features = [
    "impressions_90d",
    "clicks_90d",
    "avg_position",
    "pageviews_90d",
    "engaged_sessions_90d"
]

X = df[features].copy()
X = X.fillna(0)

print("Feature vector built successfully")
print("Rows:", len(X))
print("Features:", features)
print("\nMissing values after fill:")
print(X.isna().sum())
```

## 2. Feature notes (meaning, missing, categorical, available-when?)

*For each feature: what it means, how missing values are handled, and whether it exists BEFORE the moment you predict.*

```python
feature_notes = {
    "impressions_90d": "Search impressions observed during the historical window.",
    "clicks_90d": "Search clicks observed during the historical window.",
    "avg_position": "Average search ranking position observed historically.",
    "pageviews_90d": "Pageviews observed during the historical window.",
    "engaged_sessions_90d": "Engaged sessions observed during the historical window."
}

print("FEATURE NOTES")
for feature, meaning in feature_notes.items():
    print(f"- {feature}: {meaning}")

print("\nMissing numeric values are filled with 0.")
print("All selected features are historical signals available before the prediction decision.")
```

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

```python
forbidden_features = [
    "trend_direction",
    "trend_pct"
]

used_forbidden = [f for f in features if f in forbidden_features]

print("LEAKAGE CHECK")
print("Predictive features:", features)
print("Forbidden predictive inputs found:", used_forbidden)

assert len(used_forbidden) == 0, "Leakage detected!"

print("Leakage audit: PASS")
```
