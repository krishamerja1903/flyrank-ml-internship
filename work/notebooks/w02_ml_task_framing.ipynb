## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Ranking and Scoring.
We are calculating a risk or opportunity score for each page to rank them in a prioritized queue for the content team.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Our target is whether a page is losing momentum (observed via trend_direction == 'down').
It is a proxy because "declining traffic" is the closest measurable historical signal we have for a page that "needs a refresh."

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@K (e.g., Precision@50).
Overall accuracy does not matter as much as ensuring that if the model gives a human reviewer a top-50 list of pages to fix, the vast majority of those 50 actually require an intervention.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule is rigid and relies on hard cutoffs.
An ML model can weigh dozens of weak, interacting signals—such as competition level, search intent, scroll rates, and precise age—simultaneously to spot complex patterns of decay that a human-written if-statement would miss.
